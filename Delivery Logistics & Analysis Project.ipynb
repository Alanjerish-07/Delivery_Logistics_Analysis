# 📊 Stage 3 — Data Visualization



**Interpretation MUST with every visualization**

Focus on business story not just charts

Each chart must be in seperate cells.

```python
# Visualization - 1
# Delivery Partner Performance — Ranked Bar Chart

partner_counts = (
    df.groupby("delivery_partner")
      .size()
      .sort_values(ascending=True)
)

plt.figure(figsize=(10, 6))
partner_counts.plot(kind="barh")
plt.title("Delivery Volume by Partner")
plt.xlabel("Number of Deliveries")
plt.ylabel("Delivery Partner")
plt.tight_layout()
plt.show()
```

# Interpretation Visual-1

The bar chart compares the number of deliveries handled by each delivery partner. Longer bars indicate higher delivery volumes.

```python
# Visualization - 2
# Average Delivery Rating by Delivery Partner — Performance Ranking

partner_rating = (
    df_clean.groupby("delivery_partner")["delivery_rating"]
      .mean()
      .sort_values(ascending=True)
)

plt.figure(figsize=(10, 6))
partner_rating.plot(kind="barh", color="skyblue")
plt.title("Average Delivery Rating by Delivery Partner")
plt.xlabel("Average Rating (4 to 5)")
plt.ylabel("Delivery Partner")
plt.xlim(3.9, 5.1)  # Since minimum rating is 4.0
plt.tight_layout()
plt.show()
```

```python
# Visualization - 3
# Distance vs Delivery Cost — Scatter Plot


plot_df = df.dropna(subset=["distance_km", "delivery_cost"])

plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=plot_df,
    x="distance_km",
    y="delivery_cost",
    hue="vehicle_type",
    alpha=0.65
)
plt.title("Distance vs Delivery Cost by Vehicle")
plt.xlabel("Distance (km)")
plt.ylabel("Delivery Cost")
plt.tight_layout()
plt.show()
```

## Interpretation Visual-3

The scatter plot shows how delivery cost varies with distance. An upward pattern indicates that longer-distance deliveries tend to have higher costs. Points far from the main pattern may represent unusual deliveries.

```python
# Visualization - 4
# Vehicle Cost Efficiency — Box Plot


plot_df = df.dropna(subset=["vehicle_type", "cost_per_km"])

plt.figure(figsize=(10, 6))
sns.boxplot(
    data=plot_df,
    x="vehicle_type",
    y="cost_per_km",
    showfliers=False
)
plt.title("Cost per Kilometer by Vehicle Type")
plt.xlabel("Vehicle Type")
plt.ylabel("Cost per Kilometer")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

# Interpretation Visual-4
The box plot compares the distribution of cost per kilometer across vehicle types. The median line indicates the typical cost, while the box shows the middle 50% of observations.

```python
# visualization - 5
# Weather Impact on Delivery Ratings — Heatmap

weather_rating = (
    df_clean.dropna(subset=["weather_condition", "delivery_rating"])
      .groupby("weather_condition")[["delivery_rating"]]
      .mean()
)

plt.figure(figsize=(8, 5))
sns.heatmap(
    weather_rating,
    annot=True,
    fmt=".2f",
    cmap="YlGnBu",
    linewidths=0.5
)
plt.title("Weather Impact on Average Delivery Ratings")
plt.ylabel("Weather Condition")
plt.xlabel("Average Rating")
plt.tight_layout()
plt.show()
```

# Interpretation Visual-5

The heatmap compares delivery delay rates across weather conditions. Darker cells indicate higher delay rates when using a higher-value color scale.

```python
# viualization - 6
# Customer Rating by Delivery Mode — Violin Plot


plot_df = df.dropna(subset=["delivery_mode", "delivery_rating"])

plt.figure(figsize=(10, 6))
sns.violinplot(
    data=plot_df,
    x="delivery_mode",
    y="delivery_rating",
    inner="quartile",
    cut=0
)
plt.title("Customer Rating Distribution by Delivery Mode")
plt.xlabel("Delivery Mode")
plt.ylabel("Customer Rating")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

# Interpretation Visual-6
The violin plot illustrates the distribution of customer ratings for different delivery modes. It helps compare typical ratings and variation in customer experience.

```python
# Visualization - 7
# Regional Delivery Cost — Treemap-Style Alternative


region_cost = (
    df.groupby("region")["delivery_cost"]
      .sum()
      .sort_values()
)

plt.figure(figsize=(10, 6))
region_cost.plot(kind="barh")
plt.title("Total Delivery Cost by Region")
plt.xlabel("Total Delivery Cost")
plt.ylabel("Region")
plt.tight_layout()
plt.show()
```

# Interpretation Visal-7

The bar chart compares total delivery expenditure across regions. Regions with longer bars contribute more to total recorded delivery costs.

```python
# Visualization - 8
# Package Weight vs Delivery Cost — Regression Plot


plot_df = df.dropna(
    subset=["package_weight_kg", "delivery_cost"]
)

plt.figure(figsize=(10, 6))
sns.regplot(
    data=plot_df,
    x="package_weight_kg",
    y="delivery_cost",
    scatter_kws={"alpha": 0.35},
    line_kws={"linewidth": 2}
)
plt.title("Package Weight vs Delivery Cost")
plt.xlabel("Package Weight (kg)")
plt.ylabel("Delivery Cost")
plt.tight_layout()
plt.show()
```

# Interpretation Visual-8

The regression plot shows the relationship between package weight and delivery cost. An upward-sloping line suggests that heavier packages tend to be associated with higher costs.

```python
# Visualization - 9
# Delivery Partner Cost Efficiency — Lollipop Chart


efficiency = (
    df.groupby("delivery_partner")["cost_per_km"]
      .mean()
      .dropna()
      .sort_values(ascending=False)
)

plt.figure(figsize=(10, 6))
plt.hlines(
    y=efficiency.index,
    xmin=0,
    xmax=efficiency.values,
    linewidth=2
)
plt.scatter(efficiency.values, efficiency.index, s=65)
plt.title("Average Cost per Kilometer by Partner")
plt.xlabel("Average Cost per Kilometer")
plt.ylabel("Delivery Partner")
plt.tight_layout()
plt.show()
```

# Interpretation Visual-9

The lollipop chart ranks delivery partners according to their average cost per kilometer. Lower values indicate lower average costs per kilometer.

```python
# Visualization - 10
# Delivery Distance Categories — Stacked Bar Chart


plot_df = df.dropna(subset=["distance_km", "delivery_mode"]).copy()

plot_df["distance_category"] = pd.cut(
    plot_df["distance_km"],
    bins=[0, 5, 10, 20, float("inf")],
    labels=["Short", "Medium", "Long", "Very Long"],
    include_lowest=True
)

distance_mix = pd.crosstab(
    plot_df["distance_category"],
    plot_df["delivery_mode"],
    normalize="index"
) * 100

distance_mix.plot(
    kind="bar",
    stacked=True,
    figsize=(10, 6)
)

plt.title("Delivery Mode Mix by Distance Category")
plt.xlabel("Distance Category")
plt.ylabel("Percentage of Deliveries (%)")
plt.legend(title="Delivery Mode", bbox_to_anchor=(1.02, 1))
plt.tight_layout()
plt.show()
```

# Interpretation Visual-10

The stacked bar chart shows the percentage distribution of delivery modes within each distance category. It helps reveal whether particular modes are more common for short or long journeys.

```python
# Visualization - 11
# Numerical Feature Relationships — Correlation Heatmap
numeric_cols = [
    "distance_km",
    "package_weight_kg",
    "delivery_rating",
    "delivery_cost",
    "cost_per_km"
]

corr = df[numeric_cols].corr()

plt.figure(figsize=(9, 7))
sns.heatmap(
    corr,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0,
    vmin=-1,
    vmax=1,
    square=True
)
plt.title("Correlation Between Logistics Metrics")
plt.tight_layout()
plt.show()
```

# Interpretation Visual-11

The correlation heatmap displays the strength and direction of linear relationships between numerical variables. Values near +1 indicate strong positive correlation, while values near −1 indicate strong negative correlation.

```python
# Visualization - 12
# Delivery Partner Performance Matrix — Bubble Chart

partner_metrics = (
    df.groupby("delivery_partner")
      .agg(
          total_deliveries=("delivery_id", "count"),
          avg_rating=("delivery_rating", "mean"),
          avg_cost_km=("cost_per_km", "mean")
      )
      .dropna()
)

plt.figure(figsize=(11, 7))
sns.scatterplot(
    data=partner_metrics,
    x="avg_cost_km",
    y="avg_rating",
    size="total_deliveries",
    sizes=(100, 1000),
    alpha=0.7,
    legend="brief"
)

for partner, row in partner_metrics.iterrows():
    plt.annotate(
        str(partner),
        (row["avg_cost_km"], row["avg_rating"]),
        xytext=(5, 5),
        textcoords="offset points",
        fontsize=8
    )

plt.title("Delivery Partner Performance Matrix")
plt.xlabel("Average Cost per Kilometer")
plt.ylabel("Average Customer Rating")
plt.tight_layout()
plt.show()
```

# Interpretation Visual-12

The bubble chart compares average cost per kilometer, average customer rating, and delivery volume. Each bubble represents a partner, and bubble size represents the number of deliveries.
